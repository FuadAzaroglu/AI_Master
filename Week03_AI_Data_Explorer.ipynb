```python
# Part E
df.plot.scatter(
    x="flipper_length_mm",
    y="body_mass_g"
)
```

```python
# Part E
df["body_mass_g"].hist(
    bins=10,
    edgecolor="black"
)

plt.xlabel("Body Mass (g)")
plt.ylabel("Number of Penguins")
plt.title("Distribution of Penguin Body Mass")
plt.show()

# The histogram shows that penguin body mass varies considerably. Many
# penguins are grouped in the lower and middle weight ranges, while
# another group contains heavier penguins.
```

```python
# creating
df.plot.scatter(
    x="flipper_length_mm",
    y="body_mass_g",
    alpha=0.7  #making tiny dots or overlapped ones easily to see
)

plt.xlabel("Flipper Length (mm)")
plt.ylabel("Body Mass (g)")
plt.title("Flipper Length and Penguin Body Mass")
plt.show()

# The scatter plot shows a positive relationship where penguins with longer
# flippers generally tend to have greater body mass. However, the graph
# does not demonstrate causation. It shows an association, but it does
# not prove that longer flippers cause a penguin to become heavier.
```
