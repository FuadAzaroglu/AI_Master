```python
weighted_graph = {
    "A": {"B": 4, "C": 2},
    "B": {"D": 5},
    "C": {"D": 2, "E": 7},
    "D": {"G": 3},
    "E": {"G": 2},
    "G": {}
}

weighted_graph
```

the value 4 means traveling from A to B costs 4.
The value 2 means traveling from A to C costs 2.

```python
heuristic = {
    "A": 7,
    "B": 6,
    "C": 4,
    "D": 2,
    "E": 1,
    "G": 0
}

heuristic
```

estimated remaining cost from C to G is 4.

```python
import heapq

def astar(graph, heuristic, start, goal):

    # Each item contains:
    # (estimated total cost, actual cost, current node, path)
    frontier = [(heuristic[start], 0, start, [start])]

    # Store the lowest known cost for each node
    best_cost = {start: 0}

    while frontier:

        # Remove the item with the lowest estimated total cost
        f, g, node, path = heapq.heappop(frontier)

        # Ignore an older and more expensive route
        if g > best_cost.get(node, float("inf")):
            continue

        # Stop when the goal is reached
        if node == goal:
            return path, g

        # Look at every neighbor and its edge cost
        for neighbor, edge_cost in graph[node].items():

            # Calculate the actual cost to the neighbor
            new_g = g + edge_cost

            # Continue if this route is cheaper
            if new_g < best_cost.get(neighbor, float("inf")):

                best_cost[neighbor] = new_g

                # Estimated total cost
                new_f = new_g + heuristic[neighbor]

                # Add the new possibility to the frontier
                heapq.heappush(
                    frontier,
                    (new_f, new_g, neighbor, path + [neighbor])
                )

    return None, float("inf")
```

```python
path, cost = astar(
    weighted_graph,
    heuristic,
    "A",
    "G"
)

print("Path found:", path)
print("Total cost:", cost)
```

Why A* selected this path
A* selected A → C → D → G because it has the lowest total cost.

```python
changed_graph = {
    node: neighbors.copy()
    for node, neighbors in weighted_graph.items()
}

changed_graph["C"]["D"] = 10

changed_graph
```

I predict that A* will choose a different path. Increasing the cost from C to D makes the original route more expensive. I think A* will use C to E and then G.

```python
new_path, new_cost = astar(
    changed_graph,
    heuristic,
    "A",
    "G"
)

print("Original path:", path)
print("New path:", new_path)

print("Original cost:", cost)
print("New cost:", new_cost)
```

The result matched my prediction. The original route through D now costs 2 + 10 + 3 = 15. The route through E costs 2 + 7 + 2 = 11, so A* selects A → C → E → G.

```python
# Restore the original weighted graph
weighted_graph = {
    "A": {"B": 4, "C": 2},
    "B": {"D": 5},
    "C": {"D": 2, "E": 7},
    "D": {"G": 3},
    "E": {"G": 2},
    "G": {}
}

# Restore the original heuristic
heuristic = {
    "A": 7,
    "B": 6,
    "C": 4,
    "D": 2,
    "E": 1,
    "G": 0
}

# Make a copy so the original heuristic is not changed
changed_heuristic = heuristic.copy()

# Change C's heuristic from 4 to 8
changed_heuristic["C"] = 8

# Run A* with the changed heuristic
path, cost = astar(
    weighted_graph,
    changed_heuristic,
    "A",
    "G"
)

print("Path found:", path)
print("Total cost:", cost)
```

I think changing the heuristic for C may change the order in which A* checks the nodes. However, I think the final path may remain the same because the actual costs in the graph did not change.

BFS and A* may choose different routes because they use different information. BFS mainly searches for the route with the fewest steps, so it may prefer Route 1. A* considers the actual cost so far and the estimated cost remaining to reach the goal. Because of this, A* may prefer Route 2 even though it has one more step.

A real-world example is a navigation system. A route with fewer roads may have heavy traffic and take 30 minutes. Another route may use more roads but take only 15 minutes. In this case, fewer steps do not mean the route is better.
