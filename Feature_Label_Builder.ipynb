```python
import pandas as pd

data = pd.read_csv("train.csv")

print(data.head())
```

```python
print(data.columns)
```

```python
features = [
    "Pclass",
    "Sex",
    "Age",
    "SibSp",
    "Parch",
    "Fare"
]

X = data[features]
y = data["Survived"]

print("Features:")
print(X.head())

print("\nLabel:")
print(y.head())
```

1. Which column is being used as the target label?
The target label is the Survived column.

2.Which feature columns are used in X?
The features are Pclass, Sex, Age, SibSp, Parch, and Fare.

3. What does the target label represent?
The target label shows whether a passenger survived the Titanic disaster. A value of 1 means the passenger survived, while 0 means the passenger did not survive.

4. Why is this supervised learning?
This is supervised learning because the dataset already contains the correct result for every passenger. The model can study the passenger information together with the known survival result.

5. Is this classification or regression?
This is a classification problem because the model predicts one of two classes: survived or did not survive. It does not predict a continuous numeric amount.
