```python
from collections import deque
import heapq

graph = {
    'A': [('B', 1), ('C', 4)],
    'B': [('D', 2), ('E', 5)],
    'C': [('F', 1)],
    'D': [('G', 4)],
    'E': [('G', 1)],
    'F': [('G', 3)],
    'G': []
}

heuristic = {
    'A': 5,
    'B': 4,
    'C': 3,
    'D': 3,
    'E': 1,
    'F': 2,
    'G': 0
}
```

```python
def bfs(start, goal, graph):
    frontier = deque([(start, [start])])
    visited = set()
    expanded = 0

    while frontier:
        node, path = frontier.popleft()

        if node in visited:
            continue

        visited.add(node)
        expanded += 1

        if node == goal:
            return path, expanded

        for neighbor, cost in graph[node]:
            if neighbor not in visited:
                frontier.append((neighbor, path + [neighbor]))

    return None, expanded  # no path found


path, expanded = bfs('A', 'G', graph)

if path:
    print("Path:", path)
    print("Solution Length:", len(path) - 1)
    print("Expanded Nodes:", expanded)
else:
    print("No path found. Expanded Nodes:", expanded)
```

```python
def dfs(start, goal):
    stack = [(start, [start])]
    visited = set()
    expanded = 0

    while stack:
        node, path = stack.pop()

        if node in visited:
            continue

        visited.add(node)
        expanded += 1

        if node == goal:
            return path, expanded

        for neighbor, cost in reversed(graph[node]):
            stack.append((neighbor, path + [neighbor]))

path, expanded = dfs('A', 'G')

print("Path:", path)
print("Solution Length:", len(path)-1)
print("Expanded Nodes:", expanded)
```

```python
def ucs(start, goal, graph):
    frontier = [(0, start, [start])]
    visited = set()
    expanded = 0

    while frontier:
        cost, node, path = heapq.heappop(frontier)

        if node in visited:
            continue
        visited.add(node)
        expanded += 1

        if node == goal:
            return path, cost, expanded

        for neighbor, edge_cost in graph[node]:
            if neighbor not in visited:
                heapq.heappush(
                    frontier,
                    (cost + edge_cost, neighbor, path + [neighbor])
                )

    return None, float('inf'), expanded  # no path found


path, cost, expanded = ucs('A', 'G', graph)

print("Path:", path)
print("Path Cost:", cost)
print("Expanded Nodes:", expanded)
```

```python
def astar(start, goal, graph, heuristic):
    frontier = [(heuristic[start], 0, start, [start])]
    visited = set()
    expanded = 0

    while frontier:
        f, g, node, path = heapq.heappop(frontier)

        if node in visited:
            continue
        visited.add(node)
        expanded += 1

        if node == goal:
            return path, g, expanded

        for neighbor, edge_cost in graph[node]:
            if neighbor not in visited:
                new_g = g + edge_cost
                new_f = new_g + heuristic[neighbor]
                heapq.heappush(
                    frontier,
                    (new_f, new_g, neighbor, path + [neighbor])
                )

    return None, float('inf'), expanded  # no path found


path, cost, expanded = astar('A', 'G', graph, heuristic)

print("Path:", path)
print("Path Cost:", cost)
print("Expanded Nodes:", expanded)
```
