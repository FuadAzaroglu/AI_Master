```python
# Setup

from collections import deque
import heapq
```

```python
#Case A: BFS on a Finite Graph with Equal Costs

graph_a = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F'],
    'D': ['G'],
    'E': [],
    'F': [],
    'G': []
}

def bfs(graph, start, goal):
    frontier = deque([(start, [start])])
    visited = set()
    expanded = 0
    max_frontier = 1

    while frontier:
        max_frontier = max(max_frontier, len(frontier))

        node, path = frontier.popleft()

        if node in visited:
            continue

        visited.add(node)
        expanded += 1

        if node == goal:
            return {
                "solution": path,
                "cost": len(path)-1,
                "expanded": expanded,
                "max_frontier": max_frontier
            }

        for neighbor in graph[node]:
            if neighbor not in visited:
                frontier.append((neighbor, path + [neighbor]))

    return None

result_a = bfs(graph_a, 'A', 'G')
print(result_a)
```

```python
# Case B: DFS with a Deep Non-Goal Branch

graph_b = {
    'A': ['B', 'Goal'],
    'B': ['C'],
    'C': ['D'],
    'D': ['E'],
    'E': ['F'],
    'F': ['G'],
    'G': []
}

def dfs(graph, start, goal):
    stack = [(start, [start])]
    visited = set()
    expanded = 0
    max_frontier = 1

    while stack:
        max_frontier = max(max_frontier, len(stack))

        node, path = stack.pop()

        if node in visited:
            continue

        visited.add(node)
        expanded += 1

        if node == goal:
            return {
                "solution": path,
                "cost": len(path)-1,
                "expanded": expanded,
                "max_frontier": max_frontier
            }

        for neighbor in reversed(graph[node]):
            stack.append((neighbor, path + [neighbor]))

    return None

result_b = dfs(graph_b, 'A', 'Goal')
print(result_b)
```

```python
# Case C: Uniform-Cost Search with Unequal Costs

graph_c = {
    'A': [('B', 1), ('C', 5)],
    'B': [('D', 1)],
    'C': [('Goal', 1)],
    'D': [('Goal', 1)],
    'Goal': []
}

def ucs(graph, start, goal):
    frontier = [(0, start, [start])]
    visited = {}

    expanded = 0
    max_frontier = 1

    while frontier:
        max_frontier = max(max_frontier, len(frontier))

        cost, node, path = heapq.heappop(frontier)

        if node in visited and visited[node] <= cost:
            continue

        visited[node] = cost
        expanded += 1

        if node == goal:
            return {
                "solution": path,
                "cost": cost,
                "expanded": expanded,
                "max_frontier": max_frontier
            }

        for neighbor, edge_cost in graph[node]:
            heapq.heappush(
                frontier,
                (cost + edge_cost,
                 neighbor,
                 path + [neighbor])
            )

    return None

result_c = ucs(graph_c, 'A', 'Goal')
print(result_c)
```

```python
# Case D: A* with Admissible Heuristic

graph_d = graph_c

heuristic_good = {
    'A': 2,
    'B': 1,
    'C': 1,
    'D': 1,
    'Goal': 0
}

def astar(graph, heuristic, start, goal):

    frontier = [(heuristic[start], 0, start, [start])]
    visited = {}

    expanded = 0
    max_frontier = 1

    while frontier:
        max_frontier = max(max_frontier, len(frontier))

        f, g, node, path = heapq.heappop(frontier)

        if node in visited and visited[node] <= g:
            continue

        visited[node] = g
        expanded += 1

        if node == goal:
            return {
                "solution": path,
                "cost": g,
                "expanded": expanded,
                "max_frontier": max_frontier
            }

        for neighbor, edge_cost in graph[node]:
            new_g = g + edge_cost
            new_f = new_g + heuristic[neighbor]

            heapq.heappush(
                frontier,
                (new_f,
                 new_g,
                 neighbor,
                 path + [neighbor])
            )

    return None

result_d = astar(graph_d, heuristic_good, 'A', 'Goal')
print(result_d)
```

```python
# Case E: A* with Overestimating Heuristic

heuristic_bad = {
    'A': 10,
    'B': 10,
    'C': 1,
    'D': 10,
    'Goal': 0
}

result_e = astar(graph_d, heuristic_bad, 'A', 'Goal')
print(result_e)
```
