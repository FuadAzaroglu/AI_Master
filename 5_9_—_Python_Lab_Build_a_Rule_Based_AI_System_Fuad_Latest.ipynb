```python
facts = {
    "temperature_high",
    "cough"
}

print("Initial facts:")
print(facts)
```

The facts are stored in a Python set. At the beginning, the system knows that the patient has a high temperature and a cough. It will use these facts to find new information.


```python
rules = [
    ({"temperature_high"}, "fever"),
    ({"fever", "cough"}, "respiratory_concern"),
    ({"respiratory_concern", "oxygen_low"}, "urgent_review")
]

print("Rules:")
print(rules)
```

({"fever", "cough"}, "respiratory_concern") selecte this rule means that if the system knows the patient has a fever and a cough, it will conclude that there may be a respiratory concern. Both conditions must be known before the conclusion can be added.

```python
def forward_chain(facts, rules):
    # Make a copy of the starting facts
    known_facts = set(facts)

    # This tells the system to begin checking the rules
    changed = True

    # Continue while new facts are being found
    while changed:
        changed = False

        # Check every rule
        for conditions, conclusion in rules:

            # Check if all conditions are known
            if conditions.issubset(known_facts):

                # Add the conclusion if it is not already known
                if conclusion not in known_facts:
                    known_facts.add(conclusion)
                    print("Inferred:", conclusion)
                    changed = True

    # Return all initial and inferred facts
    return known_facts

result = forward_chain(facts, rules)

print("\nAll known facts:")
print(result)
```

Task 3:
initial facts:temperature_high and cough
inferred facts: fever and respiratory_concern

Explanation:
The system first sees the fact temperature_high. The first rule uses this fact to infer fever. The system then knows both fever and cough, so the second rule becomes true and adds respiratory_concern.
The system cannot infer urgent_review yet because oxygen_low is missing.

```python
facts.add("oxygen_low")

print("Updated initial facts:")
print(facts)

result_with_oxygen = forward_chain(facts, rules)

print("\nAll known facts:")
print(result_with_oxygen)
```

Explanation:
The result matched my prediction. The system first inferred fever and then respiratory_concern. Because oxygen_low was added, the final rule had all its required conditions and inferred urgent_review.

```python
# Restore the original facts
test_facts = {
    "temperature_high",
    "cough"
}

# Rules with one modified rule
modified_rules = [
    ({"temperature_high"}, "fever"),
    ({"fever", "cough"}, "respiratory_concern"),

    # Modified rule: oxygen_low is no longer required
    ({"respiratory_concern"}, "urgent_review")
]

modified_result = forward_chain(test_facts, modified_rules)

print("\nAll known facts:")
print(modified_result)
```

In Original rule If there is a respiratory concern and low oxygen, then an urgent review is needed. In Modified rule If there is a respiratory concern, then an urgent review is needed. As a result, system will infer urgent_review without needing oxygen_low.

Explanation: The result matched my prediction. Removing the oxygen_low condition made the rule easier to activate.

```python

({"fever"}, "monitor_temperature")

# Restore the original facts
facts_for_new_rule = {
    "temperature_high",
    "cough"
}

# Use the original rules and add one new rule
rules_with_new_rule = [
    ({"temperature_high"}, "fever"),
    ({"fever", "cough"}, "respiratory_concern"),
    ({"respiratory_concern", "oxygen_low"}, "urgent_review"),
    ({"fever"}, "monitor_temperature")
]

new_rule_result = forward_chain(
    facts_for_new_rule,
    rules_with_new_rule
)

print("\nAll known facts:")
print(new_rule_result)
```

My new rule means that if the patient has a fever, the patient’s temperature should be monitored. I chose fever as the condition because a fever may change over time. The new conclusion was monitor_temperature

```python
# Mini Rule System
it_rules = [
    (
        {"website_unavailable", "database_connection_failed"},
        "database_issue"
    ),
    (
        {"database_issue"},
        "check_database_service"
    ),
    (
        {"check_database_service", "server_cpu_high"},
        "urgent_incident"
    ),
    (
        {"urgent_incident", "recent_deployment"},
        "rollback_deployment"
    )
]
```

```python
it_facts = {
    "website_unavailable",
    "database_connection_failed",
    "server_cpu_high",
    "recent_deployment"
}

it_rules = [
    (
        {"website_unavailable", "database_connection_failed"},
        "database_issue"
    ),
    (
        {"database_issue"},
        "check_database_service"
    ),
    (
        {"check_database_service", "server_cpu_high"},
        "urgent_incident"
    ),
    (
        {"urgent_incident", "recent_deployment"},
        "rollback_deployment"
    )
]

print("Initial IT facts:")
print(it_facts)

it_result = forward_chain(it_facts, it_rules)

print("\nAll known IT facts:")
print(it_result)
```

The system starts with four facts about the website problem. Since the website is unavailable and the database connection failed, the system decides there may be a database issue. It then suggests checking the database service.
Because the server’s CPU is also high, the system identifies the problem as an urgent incident. Since there was a recent deployment, it finally suggests rolling back that deployment.

### Part 8 — Applied AI Reflection

### 1. What is the difference between an initial fact and an inferred fact?

An initial fact is information the system already has at the beginning. An inferred fact is new information the system finds by using its rules.

### 2. Why can one new fact sometimes trigger several new conclusions?

One new fact can make a rule work and produce a new conclusion. That conclusion may then activate another rule, so the system can find several new facts one after another.

### 3. What could happen if one rule contains incorrect domain knowledge?

If a rule is incorrect, the system may reach the wrong conclusion. That incorrect conclusion could also activate other rules and lead to more mistakes.

### 4. What happens if important information is missing?

If important information is missing, the system may not be able to use some rules. As a result, it may not reach a conclusion, even if that conclusion is correct in real life.

### 5. What is one advantage of a rule-based system?

One advantage is that it is easy to understand how the system makes a decision. We can look at the facts and rules to see why it reached a certain conclusion.

### 6. What is one limitation compared with a system that learns patterns from data?

A rule-based system cannot learn new patterns by itself. A person must create and update the rules, which can take a lot of time when there are many different situations.
